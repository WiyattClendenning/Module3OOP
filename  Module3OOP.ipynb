```python
class Solution:
    def firstSearch(self, arr, k):
        L = 0
        R = len(arr)-1
        
        
        while L <= R:
            M = int((L + R)/2)
            
            if arr[M] == k:
                if arr[M-1] != k:
                    return M
                else:
                    R = M-1
            elif k < arr[M]:
                R = M-1
            elif k > arr[M]:
                L = M+1
            
        return -1
```
